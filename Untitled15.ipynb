```python
#factorial using function
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

num = int(input("Enter number: "))
print("Factorial:", factorial(num))
```

```python
#prime number checker
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter number: "))

if is_prime(num):
    print("Prime Number")
else:
    print("Not Prime")
```

```python
#word frequency counter
sentence = input("Enter sentence: ")

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
#library book system
books = []

def add_book(book):
    books.append(book)

def remove_book(book):
    books.remove(book)

def search_book(book):
    if book in books:
        print("Book Found")
    else:
        print("Book Not Found")

def display_books():
    print("Books:", books)

while True:
    print("\n1.Add Book")
    print("2.Remove Book")
    print("3.Search Book")
    print("4.Display Books")
    print("5.Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        book = input("Enter book name: ")
        add_book(book)

    elif choice == 2:
        book = input("Enter book name: ")
        remove_book(book)

    elif choice == 3:
        book = input("Enter book name: ")
        search_book(book)

    elif choice == 4:
        display_books()

    elif choice == 5:
        break
```

```python
#student result analyzer
marks = []

for i in range(5):
    mark = int(input("Enter mark: "))
    marks.append(mark)

def average(marks):
    return sum(marks) / len(marks)

def passed_students(marks):
    count = 0
    for m in marks:
        if m > 40:
            count += 1
    return count

print("Average:", average(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))
print("Passed Students:", passed_students(marks))
```

```python
#password strength checker
def check_password(password):
    has_digit = False
    has_upper = False
    has_lower = False

    for ch in password:
        if ch.isdigit():
            has_digit = True
        elif ch.isupper():
            has_upper = True
        elif ch.islower():
            has_lower = True

    if len(password) >= 8 and has_digit and has_upper and has_lower:
        return "Strong Password"
    else:
        return "Weak Password"

pwd = input("Enter password: ")

print(check_password(pwd))
```

```python
#contact book application
contacts = {}

while True:
    print("\n1.Add Contact")
    print("2.Search Contact")
    print("3.Delete Contact")
    print("4.Display Contacts")
    print("5.Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        name = input("Enter name: ")
        number = input("Enter number: ")
        contacts[name] = number

    elif choice == 2:
        name = input("Enter name: ")

        if name in contacts:
            print("Number:", contacts[name])
        else:
            print("Contact not found")

    elif choice == 3:
        name = input("Enter name: ")

        if name in contacts:
            del contacts[name]

    elif choice == 4:
        print(contacts)

    elif choice == 5:
        break
```

```python
#mini banking system
balance = 1000

def deposit(amount):
    global balance
    balance += amount
    print("Deposited:", amount)

def withdraw(amount):
    global balance

    if amount <= balance:
        balance -= amount
        print("Withdrawn:", amount)
    else:
        print("Insufficient Balance")

def check_balance():
    print("Current Balance:", balance)

deposit(500)
withdraw(300)
check_balance()
```
