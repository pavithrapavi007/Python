```python
""" 
Weird or Not Weird (HackerRank classic) 
Given an integer n: 
● If n is odd → print "Weird" 
● If n is even and in range 2–5 → "Not Weird" 
● If n is even and in range 6–20 → "Weird" 
● If n is even and > 20 → "Not Weird" 
"""
n = int(input())

if n % 2 != 0:
    print("Weird")
else:
    if 2 <= n <= 5:
        print("Not Weird")
    elif 6 <= n <= 20:
        print("Weird")
    else:
        print("Not Weird")
```

```python
""" 
Triangle Validity 
Given 3 sides a, b, c, check: 
● If triangle is possible (sum of any 2 > third) 
● Print "Valid" or "Invalid"
"""
a = int(input())
b = int(input())
c = int(input())

if a + b > c and b + c > a and a + c > b:
    print("Valid")
else:
    print("Invalid")
```

```python
"""
Type of Triangle 
Given 3 sides: 
● Equilateral (all equal) 
● Isosceles (any two equal) 
● Scalene (all different)
"""
a = int(input())
b = int(input())
c = int(input())

# First check if triangle is valid
if a + b > c and b + c > a and a + c > b:
    
    if a == b == c:
        print("Equilateral")
    elif a == b or b == c or a == c:
        print("Isosceles")
    else:
        print("Scalene")
        
else:
    print("Invalid Triangle")
```

```python
"""
 Electricity Bill Calculation 
Input units consumed: 
● First 100 → ₹5/unit 
● Next 100 → ₹7/unit 
● Above 200 → ₹10/unit 
Calculate total bill.
"""
units = int(input("Enter units consumed: "))

bill = 0

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + (units - 100) * 7
else:
    bill = (100 * 5) + (100 * 7) + (units - 200) * 10

print("Total Electricity Bill: ₹", bill)
```

```python
"""
Salary Bonus Calculation 
Input salary and years of service:
≥ 5 years → 10% bonus 
 < 5 years → 5% bonus 
Print final salary.
"""
salary = float(input("Enter salary: "))
years = int(input("Enter years of service: "))

if years >= 5:
    bonus = salary * 0.10
else:
    bonus = salary * 0.05

final_salary = salary + bonus

print("Final Salary:", final_salary)
```

```python
"""
 ATM Withdrawal System 
Input: 
● Account balance 
● Withdrawal amount 
Conditions: 
● Amount must be multiple of 100 
● Balance must be sufficient 
● Minimum balance ₹500 should remain 
Print "Transaction Successful" or "Transaction Failed"
"""
 balance = float(input("Enter account balance: "))
amount = float(input("Enter withdrawal amount: "))

if amount % 100 != 0:
    print("Transaction Failed")
elif amount > balance:
    print("Transaction Failed")
elif (balance - amount) < 500:
    print("Transaction Failed")
else:
    print("Transaction Successful")
    print("Remaining Balance:", balance - amount)
```

```python
"""
 Quadratic Equation Roots Nature 
Given a, b, c, calculate discriminant D = b² - 4ac: 
● D > 0 → "Real and Distinct" 
● D = 0 → "Real and Equal" 
● D < 0 → "Imaginary" 
"""
# Input coefficients
a = float(input("Enter value of a: "))
b = float(input("Enter value of b: "))
c = float(input("Enter value of c: "))

# Calculate discriminant
D = b**2 - 4*a*c

# Check nature of roots
if D > 0:
    print("Real and Distinct")
elif D == 0:
    print("Real and Equal")
else:
    print("Imaginary")
```

```python
"""
Discount Calculator 
Input purchase amount: 
● ≥ 5000 → 20% discount 
● 2000–4999 → 10% 
● < 2000 → no discount 
Print final amount. 
"""
# Input purchase amount
amount = float(input("Enter purchase amount: "))

# Calculate discount
if amount >= 5000:
    discount = amount * 0.20
elif amount >= 2000:
    discount = amount * 0.10
else:
    discount = 0

# Final amount
final_amount = amount - discount

# Output
print("Discount:", discount)
print("Final Amount:", final_amount)
```

```python
"""
 Menu Driven Program
 Display menu: 
1. Check even/odd 
2. Check positive/negative 
3. Exit 
Take user choice and perform operation using if-elif.
"""
while True:
    print("\nMenu")
    print("1. Check even/odd")
    print("2. Check positive/negative")
    print("3. Exit")
    
    choice = int(input("Enter your choice: "))
    
    if choice == 1:
        num = int(input("Enter a number: "))
        if num % 2 == 0:
            print("Even number")
        else:
            print("Odd number")
    
    elif choice == 2:
        num = int(input("Enter a number: "))
        if num > 0:
            print("Positive number")
        elif num < 0:
            print("Negative number")
        else:
            print("Zero")
    
    elif choice == 3:
        print("Exiting program...")
        break
    
    else:
        print("Invalid choice! Please try again.")
```
